```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# (a) Generate Data & Plot in Original Space

n_samples = 2000

# Παράγουμε 2000 σημεία από ομοιόμορφη κατανομή στο διάστημα [-2, 2]
x1 = np.random.uniform(-2, 2, n_samples)
x2 = np.random.uniform(-2, 2, n_samples)

# Ορίζουμε τις κλάσεις: Αν x2^2 - x1^2 > 0 τότε κλάση +1, αλλιώς κλάση -1
y = np.where(x2**2 - x1**2 > 0, 1, -1)

# Γράφημα αρχικού χώρου
plt.figure(figsize=(6, 6))
plt.scatter(x1[y==1], x2[y==1], c='blue', marker='o', label='Class +1', alpha=0.5)
plt.scatter(x1[y==-1], x2[y==-1], c='red', marker='x', label='Class -1', alpha=0.5)
plt.title("Original Space (x1, x2)")
plt.xlabel("x1")
plt.ylabel("x2")
plt.legend()
plt.show()


# (b) Apply transformation

# Μετασχηματίζουμε τα δεδομένα υψώνοντας στο τετράγωνο
x1_prime = x1**2
x2_prime = x2**2

# Δημιουργούμε τον νέο πίνακα δεδομένων X'
X_prime = np.column_stack((x1_prime, x2_prime))


# (c) Plot in Transformed Space

plt.figure(figsize=(6, 6))
plt.scatter(x1_prime[y==1], x2_prime[y==1], c='blue', marker='o', label='Class +1', alpha=0.5)
plt.scatter(x1_prime[y==-1], x2_prime[y==-1], c='red', marker='x', label='Class -1', alpha=0.5)
plt.title("Transformed Space (x1^2, x2^2)")
plt.xlabel("x1^2")
plt.ylabel("x2^2")
plt.legend()
plt.show()

# (d) Linear model in transformed space (Least Squares)
model_trans = LinearRegression()
model_trans.fit(X_prime, y)

print("Transformed Space Least Squares Estimation")
print(f"Estimated Theta_0: {model_trans.intercept_:.4f}")
print(f"Estimated Theta_1, Theta_2: {model_trans.coef_}")
```
