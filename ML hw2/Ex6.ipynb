```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error


# (a) Generate Data
n_samples = 200

# Παράγουμε 200 τυχαία x1 και x2 από ομοιόμορφη κατανομή στο διάστημα [0, 10]
# Ακολουθώντας τη στρατηγική της Διαφάνειας 12
X1 = np.random.uniform(0, 10, n_samples)
X2 = np.random.uniform(0, 10, n_samples)

# Ο θόρυβος έχει κανονική κατανομή, μέση τιμή 0 και διακύμανση 0.05
# Η συνάρτηση normal δέχεται την τυπική απόκλιση  άρα βάζουμε τη ρίζα
eta = np.random.normal(0, np.sqrt(0.05), n_samples)

# Το πραγματικό μοντέλο που παράγει τα δεδομένα
y = 3 + 2*X1 + 1*X2 + 1*(X1*X2) + eta


# (b) Linear model in original space (X1, X2)

# Ενώνουμε τα X1 και X2 σε έναν πίνακα 200x2
X_orig = np.column_stack((X1, X2))

# Εκπαίδευση του γραμμικού μοντέλου
model_orig = LinearRegression()
model_orig.fit(X_orig, y)
print(f"Original Space")
print(f"Estimated Theta_0: {model_orig.intercept_:.4f}")
print(f"Estimated Theta_1, Theta_2: {model_orig.coef_}")


# (c) Compute MSE in original space

y_pred_orig = model_orig.predict(X_orig)
mse_orig = mean_squared_error(y, y_pred_orig)
print(f"MSE (Original): {mse_orig:.4f}\n")


# (d) Apply transformation

# Δημιουργούμε τον νέο χώρο προσθέτοντας το γινόμενο X1*X2 (πίνακας 200x3)
X_trans = np.column_stack((X1, X2, X1*X2))


# (e) Linear model in transformed space

model_trans = LinearRegression()
model_trans.fit(X_trans, y)
print(f"Transformed Space")
print(f"Estimated Theta_0: {model_trans.intercept_:.4f}")
print(f"Estimated Theta_1, Theta_2, Theta_3: {model_trans.coef_}")


# (f) Compute MSE in transformed space

y_pred_trans = model_trans.predict(X_trans)
mse_trans = mean_squared_error(y, y_pred_trans)
print(f"MSE (Transformed): {mse_trans:.4f}")
```

In the original space, the model gave a very high error  because it tried to fit a flat surface to non-linear data, ignoring the x1*x2 term.

In the transformed space, we added x1*x2 as a new third feature. Because of this, the model successfully found the correct parameters (3, 2, 1, 1) and the MSE dropped. This small remaining error is perfectly normal, as it exactly matches the variance of the random noise that we intentionally added at the beginning.
