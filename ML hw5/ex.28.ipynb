```python
import scipy.io
import numpy as np
from sklearn.mixture import GaussianMixture

# 1. Φόρτωση του αρχείου .mat
mat_data = scipy.io.loadmat('Dataset.mat')

# Ανάκτηση των πινάκων X1 και X2. 
X1 = mat_data['X1'] 
X2 = mat_data['X2']

# (a): Υπολογισμός p1(x) βάσει του X1 
# Εκπαίδευση του Gaussian Mixture Model με 1 συνιστώσα
gmm1 = GaussianMixture(n_components=1, covariance_type='full').fit(X1)

# Ορισμός των 3 σημείων δοκιμής
test_points_X1 = np.array([
    [2.01, 2.99, 3.98, 5.02],
    [20.78, -15.26, 19.38, -25.02],
    [3.08, 3.88, 4.15, 6.02]
])

# Η συνάρτηση score_samples επιστρέφει το λογάριθμο της πιθανότητας (ln p(x)). 
# Χρησιμοποιούμε np.exp() για να πάρουμε την κανονική τιμή της p(x).
pdf_values_X1 = np.exp(gmm1.score_samples(test_points_X1))

print(" (a): Τιμές p1(x)")
for i, val in enumerate(pdf_values_X1):
    print(f"Σημείο x{i+1}: {val:.6e}")


# (b): Υπολογισμός p2(x) βάσει του X2 
# Εκπαίδευση του δεύτερου μοντέλου στο X2
gmm2 = GaussianMixture(n_components=1, covariance_type='full').fit(X2)

# Ορισμός των 4 σημείων δοκιμής
test_points_X2 = np.array([
    [0.05, 0.15, -0.12, -0.08],
    [7.18, 7.98, 9.12, 9.94],
    [3.48, 4.01, 4.55, 4.96],
    [20.78, -15.26, 19.38, -25.02]
])

pdf_values_X2 = np.exp(gmm2.score_samples(test_points_X2))

print("\n (b): Τιμές p2(x)")
for i, val in enumerate(pdf_values_X2):
    print(f"Σημείο x{i+1}: {val:.6e}")
```
